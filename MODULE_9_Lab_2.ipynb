# Data preparation

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Load the iris dataset
iris = load_iris()
X = iris['data']
y = iris['target']
names = iris['target_names']
feature_names = iris['feature_names']

# Scale data to have mean 0 and variance 1
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split the data set into training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=2)
```

# Visualising dataset

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
for target, target_name in enumerate(names):
    X_plot = X[y == target]
    ax1.plot(X_plot[:, 0], X_plot[:, 1],
             linestyle='none',
             marker='o',
             label=target_name)
ax1.set_xlabel(feature_names[0])
ax1.set_ylabel(feature_names[1])
ax1.axis('equal')
ax1.legend();

for target, target_name in enumerate(names):
    X_plot = X[y == target]
    ax2.plot(X_plot[:, 2], X_plot[:, 3],
             linestyle='none',
             marker='o',
             label=target_name)
ax2.set_xlabel(feature_names[2])
ax2.set_ylabel(feature_names[3])
ax2.axis('equal')
ax2.legend();
```

# MLP for multiclass classification

```python
import torch
import torch.nn.functional as F
import torch.nn as nn
from torch.autograd import Variable
import tqdm
```

```python
# Defining the model architecture
class Model(nn.Module):
    def __init__(self, input_dim):
        super(Model, self).__init__()
        self.layer1 = nn.Linear(input_dim, 50)
        self.layer2 = nn.Linear(50, 50)
        self.layer3 = nn.Linear(50, 3)

    def forward(self, x):
        x = F.relu(self.layer1(x))
        x = F.relu(self.layer2(x))
        x = F.softmax(self.layer3(x), dim=1)
        return x
```

```python
# Instantiating the model, using Adam optimiser, and Cross Entropy Loss, which is quite commonlu used for classification tasks.
model     = Model(X_train.shape[1])
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
loss_fn   = nn.CrossEntropyLoss()
```

# Train the model

```python
# Train for 100 epochs
EPOCHS  = 100
X_train = torch.from_numpy(X_train).float()
X_test= torch.from_numpy(X_test).float()
y_test = torch.from_numpy(y_test)
y_train = torch.from_numpy(y_train)


loss_list     = np.zeros((EPOCHS,))
accuracy_list = np.zeros((EPOCHS,))

for epoch in tqdm.trange(EPOCHS):
    y_pred = model(X_train)
    loss = loss_fn(y_pred, y_train)
    loss_list[epoch] = loss.item()

    # Zero gradients
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    with torch.no_grad():
        y_pred = model(X_test)
        correct = (torch.argmax(y_pred, dim=1) == y_test).type(torch.FloatTensor)
        accuracy_list[epoch] = correct.mean()
```

# Plot training progress

```python
fig, (ax1, ax2) = plt.subplots(2, figsize=(12, 6), sharex=True)

ax1.plot(accuracy_list)
ax1.set_ylabel("validation accuracy")
ax2.plot(loss_list)
ax2.set_ylabel("validation loss")
ax2.set_xlabel("epochs");
```

1. Changing the Number of Hidden Layers:
Effect on Model: The number of hidden layers in a neural network determines the depth of the model. Adding more layers allows the model to learn more complex patterns and represent more intricate relationships in the data.

Increased Depth:

More hidden layers can help the network capture more complex relationships, making it more powerful, especially for tasks like image recognition or natural language processing.

However, after a certain point, adding too many layers may lead to overfitting or vanishing/exploding gradients, making the model harder to train.

Decreased Depth:

Fewer hidden layers mean the network is less capable of learning complex patterns, but it may be simpler to train and less prone to overfitting.

2. Changing the Number of Neurons in the Hidden Layer:
Effect on Model: The number of neurons in a hidden layer controls the width of that layer. More neurons provide the model with more capacity to learn from the data.

Increased Neurons:

More neurons can lead to a more expressive model that can learn more from the data.

However, adding too many neurons may also result in overfitting (if the model is too flexible) or increased computation time, especially with large datasets.

Decreased Neurons:

Fewer neurons might make the model simpler, leading to underfitting, where it can't capture the patterns in the data as well.

3. Using a Different Activation Function:
Effect on Model: The activation function determines the output of each neuron and introduces non-linearity to the model. Different activation functions can have a significant impact on how well the model learns.

Common Activation Functions:

ReLU (Rectified Linear Unit):

It's widely used in hidden layers due to its simplicity and efficiency. It helps with the vanishing gradient problem (gradients don't vanish as quickly as with other functions).

Effect: It allows faster training, but may cause "dying ReLU" where some neurons stop learning.

Sigmoid:

Outputs values between 0 and 1. It’s mainly used in binary classification models.

Effect: It can suffer from vanishing gradients, especially in deep networks.

Tanh (Hyperbolic Tangent):

Outputs values between -1 and 1. It’s similar to the sigmoid but better centered around 0.

Effect: Like sigmoid, it can still suffer from vanishing gradients.

Leaky ReLU:

It's a variation of ReLU that allows small negative values when the input is less than 0, which can help mitigate the dying ReLU problem.

Effect: It often provides better performance than ReLU in deeper networks.

Softmax:

Often used in the output layer for multi-class classification tasks. It converts raw output scores into probability distributions.

Effect: It’s helpful for classification tasks with multiple classes, as it ensures the outputs are in the [0, 1] range and sum to 1.
